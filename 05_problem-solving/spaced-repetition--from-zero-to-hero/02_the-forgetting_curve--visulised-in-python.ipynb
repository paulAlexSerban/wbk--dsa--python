## The Forgetting Curve - Visualised in Python

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Ebbinghaus forgetting curve: R = e^(-t/S)
# R = retrievability (0-1), t = time elapsed, S = stability (days)
```

```python
def forgetting_curve(t, stability):
    """
    Returns the probability of recall at time t (days),
    given a memory starbility (days)
    """

    return np.exp(-t / stability)

```

```python
t = np.linspace(0, 30, 300) # 0 to 30 days

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: different stabilities

for S in [1, 3, 7, 14]:
    axes[0].plot(t, forgetting_curve(t, S), label=f"S={S} days")

axes[0].axhline(0.9, color='red', linestyle='--', alpha=0.5, label="90% retention target")
axes[0].set_xlabel("Days since learning")
axes[0].set_ylabel("Probability of recall")
axes[0].set_title("Forgetting Curves at Different Stabilities")
axes[0].legend()
axes[0].grid(True, alpha=0.3)
review_days = [0, 1, 3, 8, 21]
stability = 1.0
times, retentions, stabilities = [], [], []

for i, day in enumerate(review_days):
    # Decay since last review
    if i > 0:
        gap = review_days[i] - review_days[i-1]
        t_plot = np.linspace(review_days[i-1], day, 50)
        R_plot = [forgetting_curve(d - review_days[i-1], stability) for d in t_plot]
        axes[1].plot(t_plot, R_plot, 'b-', alpha=0.6)
    # review bosts stability
    stability *= 2.5 # simplified: each review multiples stability

axes[1].set_xlabel("Days")
axes[1].set_ylabel("Retention")
axes[1].set_title("Retention with Sapced Reviews (stability grows)")

for d in review_days:
    axes[1].axvline(d, color='green', linestyle=':', alpha=0.7)

axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

```

```python
print(f"Key insight: reviewing at 90% retention doubles stability each time")
print(f"After 5 reviews: interval grow from 1 -> {1 * 2.5**5:.0f} days")
```
